## Etapa 2 - Identificar as variáveis


### Variável independente
A variável independente é o **mês (`mes`)**, pois representa o período analisado e é utilizada como entrada para analisar a variação dos custos.

$$x = mes$$

### Variável dependente

A variável dependente é o **custo de transporte (`custo_transporte`)**, pois seu valor varia de acordo com o mês analisado.

$$y = custo\_transporte$$

### Representação das variáveis

- **`mes` (x):** representa o número do mês em que o custo de transporte foi registrado.
- **`custo_transporte` (y):** representa o custo total de transporte da empresa em cada mês, em reais (R$).

## Etapa 3 - Gráfico de dispersão

O gráfico de dispersão é utilizado para representar a relação entre as duas variáveis. O eixo x representa os meses e o eixo y representa os custos de transporte, permitindo visualizar como os custos variam ao longo do tempo.

```python
import matplotlib.pyplot as plt

plt.scatter(dados["mes"], dados["custo_transporte"])

plt.title("Custo de Transporte por Mês")
plt.xlabel("Mês")
plt.ylabel("Custo de transporte (R$)")
plt.grid()

plt.show()
```

### Análise do comportamento dos dados

Os custos de transporte aumentam ao longo dos meses. Apesar de algumas variações, é possível perceber uma tendência de crescimento nos valores.

## Etapa 4 - Função afim

A forma geral de uma função afim é:

$$
f(x) = ax + b
$$

Para encontrar uma função que represente aproximadamente o comportamento dos custos, será utilizada uma reta de tendência calculada a partir dos dados apresentados na base.

### Cálculo dos coeficientes

O método `polyfit` da biblioteca NumPy é utilizado para calcular os coeficientes `a` e `b` de uma reta que melhor representa a tendência dos dados.


### Cálculo dos coeficientes

O método `polyfit` é utilizado para calcular os coeficientes `a` e `b` da reta que melhor representa a tendência dos dados.

```python
import numpy as np

x = dados["mes"]
y = dados["custo_transporte"]

a, b = np.polyfit(x, y, 1)

print(f"Coeficiente a = {a:.2f}")
print(f"Coeficiente b = {b:.2f}")
```

### Função encontrada

A partir dos coeficientes calculados, temos:

$$
a \approx 1791{,}61
$$

$$
b \approx 42204{,}55
$$

Substituindo esses valores na forma geral da função afim, temos:

$$
\boxed{f(x) \approx 1791{,}61x + 42204{,}55}
$$

## Etapa 5 - Interpretar a função

Função encontrada:

$$
f(x) \approx 1791{,}61x + 42204{,}55
$$


#### a) O que representa o coeficiente a no contexto do problema?

O coeficiente `a`, aproximadamente 1791,61, representa a taxa de variação média estimada do custo de transporte por mês. Portanto, de acordo com a função, o custo aumenta aproximadamente R$ 1.791,61 por mês.


#### b) O que representa o coeficiente b?

O coeficiente `b`, aproximadamente 42204,55, representa o valor estimado do custo de transporte quando `x = 0`, ou seja, o ponto em que a reta de tendência intercepta o eixo y. Como os dados analisados começam no mês 1, esse valor não corresponde a um custo observado na base.


#### c) O valor de a indica crescimento ou redução dos custos?

Como o coeficiente `a` é positivo, a função indica uma tendência de crescimento dos custos de transporte ao longo dos meses.

## Etapa 6 - Representar a função graficamente

Nesta etapa, os dados reais serão representados junto com a reta obtida na Etapa 4. Assim, é possível visualizar como a função se ajusta aos valores observados e identificar a tendência dos custos ao longo dos meses.

```python
import matplotlib.pyplot as plt
import numpy as np

# Valores de x para representar a reta
x_funcao = np.linspace(1, 15, 100)

# Valores de y calculados pela função
y_funcao = a * x_funcao + b

# Dados reais
plt.scatter(dados["mes"], dados["custo_transporte"], label="Dados reais")

# Reta da função
plt.plot(x_funcao, y_funcao, label="Função encontrada")

plt.title("Custo de Transporte por Mês")
plt.xlabel("Mês")
plt.ylabel("Custo de transporte (R$)")
plt.legend()
plt.grid()

plt.show()
```

O gráfico mostra que os custos de transporte tendem a aumentar ao longo dos meses, e a reta acompanha esse crescimento.

## Etapa 7 - Previsões

Utilizando a função encontrada na Etapa 4, será estimado o custo de transporte para os meses 13, 14 e 15.

$$
f(x) \approx 1791{,}61x + 42204{,}55
$$

```python
# Função para previsão
def prever_custo(mes):
    return a * mes + b

# Previsões para os meses 13, 14 e 15
for mes in [13, 14, 15]:
    custo = prever_custo(mes)
    print(f"Mês {mes}: R$ {custo:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
```

**Mês 13:**

$$

$$

$$

**Mês 14:**

$$

$$

**Mês 15:**

$$
